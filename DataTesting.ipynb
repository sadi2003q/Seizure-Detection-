```python
import pandas as pd

df = pd.read_csv("Epileptic_Sub_Eve_ID.csv")   # use your actual path

print(df.shape)
print(df.columns[:12].tolist())     # first columns: look for id / time / file columns
print(df.columns[-12:].tolist())    # last columns

# 1) Columns that are NOT the 924 features
feat_like = [c for c in df.columns if any(k in c.lower() for k in
             ["ch", "mean", "std", "psd", "band", "entropy", "spike", "lif"])]
meta = [c for c in df.columns if c not in feat_like]
print("Likely metadata columns:", meta)

# 2) Anything that looks like time / order / recording
for c in df.columns:
    if any(k in c.lower() for k in ["time", "start", "end", "onset", "offset",
                                     "idx", "index", "window", "file", "record", "edf", "sec"]):
        print("Candidate:", c, df[c].head(3).tolist())
```
